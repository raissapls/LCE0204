```python
import pandas as pd
df = pd.read_excel('dados_semana02.xlsx')
```

```python
import matplotlib.pyplot as plt

# Count the occurrences of each category in the 'Tratamento' column
treatment_counts = df['Tratamento'].value_counts()

# Create a pie chart
fig = plt.figure(figsize=(8, 8))
plt.pie(treatment_counts, labels=treatment_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribuição da categoria Tratamentos')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.legend(title="Tratamentos") # Add a legend
plt.show()
```

### Análise por Tratamento para Variáveis Numéricas

Agora vamos gerar gráficos de barras para cada variável numérica, mostrando a média para cada tratamento (T1, T2, T3, T4).

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Identify numerical columns, excluding 'Tratamento'
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()

# The 'Altura_cm' column appears to have mixed types, including dates.
# Let's convert it to numeric, coercing errors to NaN.
df['Altura_cm'] = pd.to_numeric(df['Altura_cm'], errors='coerce')

# If 'Altura_cm' was not initially identified as numeric due to mixed types,
# ensure it's added to the list if it contains meaningful numeric data after conversion.
if 'Altura_cm' not in numerical_cols:
    numerical_cols.append('Altura_cm')

# Remove 'Tratamento' from numerical_cols if it somehow got in there
if 'Tratamento' in numerical_cols:
    numerical_cols.remove('Tratamento')

# Handle the column with a space in its name if it exists and is numerical
if 'Massa _seca_R_g' in df.columns and 'Massa _seca_R_g' not in numerical_cols:
    # Ensure it's treated as numerical
    df['Massa _seca_R_g'] = pd.to_numeric(df['Massa _seca_R_g'], errors='coerce')
    numerical_cols.append('Massa _seca_R_g')

# Create a bar chart for each numerical variable
for col in numerical_cols:
    if col == 'Tratamento': # Skip 'Tratamento' as it's the grouping variable
        continue

    # Group by 'Tratamento' and calculate the mean of the current numerical column
    grouped_data = df.groupby('Tratamento')[col].mean().reset_index()

    fig = plt.figure(figsize=(10, 6))
    sns.barplot(x='Tratamento', y=col, data=grouped_data, palette='viridis')
    plt.title(f'Média de {col} por Tratamento')
    plt.xlabel('Tratamento')

    # Conditionally add '(g)' to the ylabel for specific columns
    if col in ['Massa_seca_PA_g', 'Massa _seca_R_g']:
        plt.ylabel(f'Média de {col} (g)')
    else:
        plt.ylabel(f'Média de {col}')

    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()
```

### Histogramas para Variáveis Numéricas

Agora vamos criar histogramas para cada variável numérica para entender a distribuição dos dados, com títulos e rótulos de eixo claros.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Identify numerical columns, excluding 'Tratamento' as it's a categorical grouping variable
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()

# Ensure 'Altura_cm' is treated as numeric
df['Altura_cm'] = pd.to_numeric(df['Altura_cm'], errors='coerce')
if 'Altura_cm' not in numerical_cols:
    numerical_cols.append('Altura_cm')

# Ensure 'Massa _seca_R_g' is treated as numeric
if 'Massa _seca_R_g' in df.columns and 'Massa _seca_R_g' not in numerical_cols:
    df['Massa _seca_R_g'] = pd.to_numeric(df['Massa _seca_R_g'], errors='coerce')
    numerical_cols.append('Massa _seca_R_g')

# Remove 'Tratamento' from numerical_cols if it somehow got in there
if 'Tratamento' in numerical_cols:
    numerical_cols.remove('Tratamento')

# Generate a histogram for each numerical variable
for col in numerical_cols:
    if col == 'Tratamento': # Skip 'Tratamento' as it's the grouping variable
        continue

    fig = plt.figure(figsize=(10, 6))
    sns.histplot(data=df, x=col, kde=True, bins=10, palette='viridis') # Using kde for density estimation
    plt.title(f'Histograma de {col}')

    # Conditionally add '(g)' to the xlabel for specific columns
    if col in ['Massa_seca_PA_g', 'Massa _seca_R_g']:
        plt.xlabel(f'{col} (g)')
    else:
        plt.xlabel(col)

    plt.ylabel('Frequência')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()
```
